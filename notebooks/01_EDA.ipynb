# 01 — Exploratory Data Analysis

Load the raw loan dataset (`Loan_default.csv`) and explore distributions, missing values, correlations, and target class balance.

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries Imported Successfully")
```

```python
df = pd.read_csv("../data/raw/Loan_default.csv")

print("=" * 50)
print("Dataset Loaded Successfully")
print("=" * 50)
print("Shape :", df.shape)

df.head()
```

```python
df.info()
```

```python
df.describe().T
```

```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": round(df.isnull().mean() * 100, 2)
})

missing = missing[missing["Missing Values"] > 0]

if missing.empty:
    print("No missing values found!")
else:
    display(missing.sort_values("Percentage", ascending=False))
```

```python
print("Duplicate Rows :", df.duplicated().sum())
```

```python
df.columns.tolist()
```

```python
data_types = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str)
})

data_types
```

```python
# Target column in Loan_default.csv is 'Defaulted?'
TARGET = "Defaulted?"

df[TARGET].value_counts()
```

```python
round(df[TARGET].value_counts(normalize=True) * 100, 2)
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x=TARGET)

plt.title("Loan Default Distribution")
plt.xlabel("Defaulted?")
plt.ylabel("Count")

plt.tight_layout()
plt.savefig("../data/processed/eda_default_distribution.png", dpi=80)
plt.show()
print("Plot saved.")
```

```python
# All features except Index (ID) and target
num_features = [col for col in df.select_dtypes(include=["int64", "float64"]).columns
                if col not in ["Index", TARGET]]

cat_features = df.select_dtypes(include="object").columns.tolist()

print("Numerical Features:", num_features)
print("Categorical Features:", cat_features)
```

```python
for col in num_features:

    plt.figure(figsize=(6, 4))
    sns.histplot(data=df, x=col, kde=True, bins=30)
    plt.title(f"Distribution of {col}")
    plt.tight_layout()
    plt.show()
```

```python
for col in num_features:

    plt.figure(figsize=(8, 2))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.tight_layout()
    plt.show()
```

```python
plt.figure(figsize=(10, 7))

corr = df[num_features + [TARGET]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
for col in num_features:

    plt.figure(figsize=(6, 4))
    sns.boxplot(data=df, x=TARGET, y=col)
    plt.title(f"{col} vs Defaulted?")
    plt.tight_layout()
    plt.show()
```
